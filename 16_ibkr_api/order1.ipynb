## Limit order on AMZN

```python
# get current market price to choose a sensible limit price
c2 = Stock('AMZN', 'SMART', 'USD')
ib.qualifyContracts(c2)

ticker = ib.reqMktData(c2, '', False, False)
ib.sleep(2)
print('last:', ticker.last, 'bid:', ticker.bid, 'ask:', ticker.ask)
```

```python
c2 = Stock('AMZN', 'SMART', 'USD')
# limit order: BUY 1 AMZN, only fills at limit_price or better
limit_price = 230.50   # set your price here (e.g. ticker.bid or below)

o2 = LimitOrder('BUY', 1, limit_price)
trade = ib.placeOrder(c2, o2)
trade
```

```python
# same thing using the generic Order class
o3 = Order(action='BUY', totalQuantity=1, orderType='LMT', lmtPrice=limit_price, tif='DAY')
ib.placeOrder(c2, o3)
```

```python
# check order status; cancel with ib.cancelOrder(o2) if needed
ib.sleep(1)
print(trade.orderStatus.status)
ib.openOrders()
```

## stop order and stop-limit order using Order class

```python
# stop order (STP) using the generic Order class
# once price hits stop_price, a market order is submitted
c3 = Stock('META', 'SMART', 'USD')
c3 = ib.qualifyContracts(c3)[0]

stop_price = 300

o4 = Order(action='SELL', totalQuantity=1, orderType='STP', auxPrice=stop_price, tif='DAY')
ib.placeOrder(c3, o4)
```

```python
# stop-limit order (STP LMT) using the generic Order class
# once price hits stop_price, a limit order at limit_price is submitted
stop_price = 300
limit_price = 298

o5 = Order(action='SELL', totalQuantity=1, orderType='STP LMT', auxPrice=stop_price, lmtPrice=limit_price, tif='DAY')
ib.placeOrder(c3, o5)
```
