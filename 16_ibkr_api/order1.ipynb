## stop order and stop-limit order using Order class

```python
# stop order (STP) using the generic Order class
# once price hits stop_price, a market order is submitted
c3 = Stock('META', 'SMART', 'USD')
c3 = ib.qualifyContracts(c3)[0]

stop_price = 300

o4 = Order(action='SELL', totalQuantity=1, orderType='STP', auxPrice=stop_price, tif='DAY')
ib.placeOrder(c3, o4)
```

cover order (entry + compulsory stop-loss, IBKR has no native CO type like Zerodha/Upstox — this is the parent/child equivalent)

```python
c5 = Stock('NVDA', 'SMART', 'USD')
c5 = ib.qualifyContracts(c5)[0]

qty = 1
stop_loss_price = 170  # mandatory stop-loss, like a Zerodha/Upstox Cover Order

# entry leg: market order, held back until the stop-loss leg is attached
entry = Order(action='BUY', totalQuantity=qty, orderType='MKT', transmit=False)
entry.orderId = ib.client.getReqId()

# stop-loss leg: linked to the entry via parentId, transmit=True sends both together
stop_loss = Order(action='SELL', totalQuantity=qty, orderType='STP', auxPrice=stop_loss_price,
                   parentId=entry.orderId, transmit=True)
stop_loss.orderId = ib.client.getReqId()

ib.placeOrder(c5, entry)
ib.placeOrder(c5, stop_loss)
```

bracket order (entry + take-profit + stop-loss, all linked)

```python
c6 = Stock('NVDA', 'SMART', 'USD')
c6 = ib.qualifyContracts(c6)[0]

qty = 1
limit_price = 175          # entry limit price
take_profit_price = 185    # exits with profit if price rises to this level
stop_loss_price = 168      # exits with loss if price falls to this level

# ib.bracketOrder builds 3 linked orders: parent (entry) + takeProfit + stopLoss,
# with parentId/transmit already wired so only the last one needs transmit=True
bracket = ib.bracketOrder('BUY', qty, limitPrice=limit_price,
                           takeProfitPrice=take_profit_price,
                           stopLossPrice=stop_loss_price)

for o in bracket:
    ib.placeOrder(c6, o)
```

```python
# same bracket order, built manually using the generic Order class
c7 = Stock('QQQ', 'SMART', 'USD')
c7 = ib.qualifyContracts(c7)[0]

qty = 1
limit_price = 175          # entry limit price
take_profit_price = 185
stop_loss_price = 168

parent = Order(action='BUY', totalQuantity=qty, orderType='LMT', lmtPrice=limit_price,
                transmit=False)
parent.orderId = ib.client.getReqId()

take_profit = Order(action='SELL', totalQuantity=qty, orderType='LMT', lmtPrice=take_profit_price,
                     parentId=parent.orderId, transmit=False)
take_profit.orderId = ib.client.getReqId()

stop_loss = Order(action='SELL', totalQuantity=qty, orderType='STP', auxPrice=stop_loss_price,
                   parentId=parent.orderId, transmit=True)
stop_loss.orderId = ib.client.getReqId()

ib.placeOrder(c7, parent)
ib.placeOrder(c7, take_profit)
ib.placeOrder(c7, stop_loss)
```

get all orders as a DataFrame

```python
import pandas as pd

# ib.trades() covers this session's orders; ib.reqAllOpenOrders() pulls in open orders
# placed from other clients/TWS too
ib.reqAllOpenOrders()
ib.sleep(1)

rows = []
for t in ib.trades():
    rows.append({
        'orderId': t.order.orderId,
        'symbol': t.contract.symbol,
        'action': t.order.action,
        'orderType': t.order.orderType,
        'totalQuantity': t.order.totalQuantity,
        'lmtPrice': t.order.lmtPrice,
        'auxPrice': t.order.auxPrice,
        'status': t.orderStatus.status,
        'filled': t.orderStatus.filled,
        'remaining': t.orderStatus.remaining,
        'avgFillPrice': t.orderStatus.avgFillPrice,
    })

orders_df = pd.DataFrame(rows)
orders_df
```
