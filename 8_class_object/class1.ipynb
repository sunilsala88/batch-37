```python
class Circle:
    pi=3.14
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        area=self.pi*(self.radius**2)
        return area

    def circumference(self):
        circumference=2*self.pi*self.radius
        return circumference


c1=Circle(20)
print(c1.area())
print(c1.circumference())
```
