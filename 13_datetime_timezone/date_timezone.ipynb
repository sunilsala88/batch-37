```python
import yfinance as yf
import pandas as pd

data = yf.download('TSLA', period='60d', interval='15m', multi_level_index=False)

prev_month = pd.Timestamp.now().to_period('M') - 1
prev_high = data.loc[data.index.to_period('M') == prev_month, 'High'].max()

print(f"{prev_month} high: {prev_high:.2f}")
```
